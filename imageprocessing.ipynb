```python
# ==============================================================================
# Complete Image Processing Pipeline for Google Colab
# ==============================================================================

import cv2
import requests
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# 1. Download a standard sample image
url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
response = requests.get(url)
with open("sample.jpg", "wb") as f:
    f.write(response.content)

# 2. Load the image
img = cv2.imread('sample.jpg')

# 3. Process the image
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)                     # BGR to RGB
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)                    # Grayscale
blurred_img = cv2.GaussianBlur(gray_img, (5, 5), 0)                 # Gaussian Blur
edges = cv2.Canny(blurred_img, threshold1=100, threshold2=200)      # Canny Edges
_, threshold_img = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY) # Thresholding

# 4. Display results using Matplotlib
plt.figure(figsize=(12, 8))
titles = ['Original (RGB)', 'Grayscale', 'Gaussian Blur', 'Canny Edges', 'Binary Threshold']
images = [img_rgb, gray_img, blurred_img, edges, threshold_img]

for i in range(5):
    plt.subplot(2, 3, i + 1)
    if len(images[i].shape) == 2:
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()

# 5. Native Colab display preview
print("\nNative Colab OpenCV Preview:")
cv2_imshow(img)
```
